```python
import numpy as np
from sklearn.linear_model import SGDClassifier
```

```python
X = np.array([
    [2,1],
    [1,2],
    [-1,-1],
    [-2,-1],
    [1,-1]
])
y = np.array([1,1,-1,-1,1])
```

```python
clf = SGDClassifier(
    learning_rate='constant',
    eta0=0.1,
    alpha=0.0,
    max_iter=1,
    shuffle=False,
    fit_intercept=True,
    tol=None,
    random_state=42
)
```

```python
clf.fit(X,y)
```

```python
w1,w2 = clf.coef_[0]
print(f" w1 = {w1:.4f}")
print(f" w2 = {w2:.4f}")
```

```python
b = clf.intercept_[0]
print(f"intercept: {b}")
```

```python
clf.score(X,y)
```

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("iris.csv")
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
X = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]
y = df['species']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,train_size=0.8,random_state=42)
```

```python
model = SGDClassifier()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict([[5.1	,3.5,	1.4	,0.2	]])
```
