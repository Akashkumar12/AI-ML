```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeClassifier,export_text,plot_tree
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('iris_synthetic.csv')
```

```python
df.head()
```

```python
df.columns
```

```python
features = ['petal_length', 'petal_width']
```

```python
train = df[df.split == 'train']
test = df[df.split == 'test']
```

```python
X_train,y_train = train[features],train['label']
X_test,y_test = test[features],test['label']
```

```python
for depth in range(1,13):
  model = DecisionTreeClassifier(criterion='entropy',max_depth=depth,random_state=0)
  model.fit(X_train,y_train)
  train_acc = model.score(X_train,y_train)
  test_acc = model.score(X_test,y_test)
  print(f"Depth: {depth}, Train Acc: {train_acc:5.0%}, Test Acc: {test_acc:5.0%}")
```

```python
m1 = DecisionTreeClassifier(criterion='entropy',max_depth=3,random_state=0)
m1.fit(X_train,y_train)
print(m1.score(X_train,y_train))
print(m1.score(X_test,y_test))
```

```python
import matplotlib.pyplot as plt
depths = range(1,13)
train_scores ,test_scores =[],[]
for d in depths:
  model = DecisionTreeClassifier(criterion='entropy',max_depth=d,random_state=0)
  model.fit(X_train,y_train)
  train_scores.append(model.score(X_train,y_train))
  test_scores.append(model.score(X_test,y_test))
plt.plot(depths,train_scores,"-o",label='train_scores')
plt.plot(depths,test_scores,"--o",label='test_scores')
plt.xlabel("Depth")
plt.ylabel("Accuracy")
plt.legend()
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
grid = {
    "max_depth":[2,3,4,5,6,None],
    "min_samples_leaf":[1,5,10,15],
    "min_samples_split":[2,10,20]
}
```

```python
search = GridSearchCV(DecisionTreeClassifier(criterion='entropy',random_state=0),grid,cv=5)
search.fit(X_train,y_train)
```

```python
best = search.best_estimator_
print("Best Settings for our Dt to avaoid overfitting :",search.best_params_)
print("Train Score :",best.score(X_train,y_train))
print("Test Score :",best.score(X_test,y_test))
```

```python
import optuna
from sklearn.model_selection import cross_val_score
optuna.logging.set_verbosity(optuna.logging.WARNING)
def objective(trial):
  params = {
      "max_depth":trial.suggest_int("max_depth",2,12),
      "min_samples_leaf":trial.suggest_int("min_samples_leaf",1,20),
      "min_samples_split":trial.suggest_int("min_samples_split",2,20)
  }

  model = DecisionTreeClassifier(criterion='entropy',random_state=0,**params)
  score = cross_val_score(model,X_train,y_train,cv=5).mean()
  return cross_val_score(model,X_train,y_train,cv=5).mean()
```

```python
study = optuna.create_study(direction='maximize',sampler=optuna.samplers.TPESampler(seed=42))
study.optimize(objective,n_trials=100)
```

```python
best = DecisionTreeClassifier(criterion='entropy',random_state=0,**study.best_params)
best.fit(X_train,y_train)
print("Best Settings for our Dt to avaoid overfitting :",study.best_value)
```
