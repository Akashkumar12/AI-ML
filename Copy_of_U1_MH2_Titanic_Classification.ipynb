### 3.1 Working on the "Cabin" column (2 Marks)
Find unique entries in the Cabin column. We can label all passengers in two categories having a cabin or not. Check the data type(use: type) of each entry of the Cabin. Convert a string data type into '1' i.e. passengers with cabin and others into '0' i.e. passengers without cabin.  Write a function for the above operation and apply it to the cabin column and create another column with the name " Has_cabin" containing only 0 or 1 entries.





```
print('Unique entries in Cabin column (X_train):')
print(X_train['Cabin'].unique())
print('\nData type of Cabin column (X_train):')
print(X_train['Cabin'].dtype)
```

```
def has_cabin(cabin_entry):
    if pd.isna(cabin_entry):
        return 0  # No cabin
    else:
        return 1  # Has a cabin

X_train['Has_cabin'] = X_train['Cabin'].apply(has_cabin)
print('\nX_train with new Has_cabin column:')
display(X_train[['Cabin', 'Has_cabin']].head())
```

### 3.4 Working on the "Age" column (2 Marks)
find the number of NaN entries in the age column and their row index. Calculate the mean, Standard deviation of the Age column and check the distribution of the age column.We can fill the missing values with randomly generated integer values between (mean+Standard deviation, mean-Standard deviation). Use : np.isnan; np.random.randint; concept of slicing dataframe. Convert the age column as an integer data type.



```
nan_age_count = X_train['Age'].isnull().sum()
nan_age_indices = X_train[X_train['Age'].isnull()].index.tolist()
print(f"Number of NaN entries in Age column: {nan_age_count}")
print(f"Row indices of NaN entries in Age column: {nan_age_indices}")
```

```
age_mean = X_train['Age'].mean()
age_std = X_train['Age'].std()
print(f"Mean of Age column: {age_mean:.2f}")
print(f"Standard deviation of Age column: {age_std:.2f}")
print("\nDistribution of Age column:")
display(X_train['Age'].describe())
```

```
import numpy as np

# Calculate bounds for random imputation
lower_bound = int(age_mean - age_std)
upper_bound = int(age_mean + age_std)

# Get indices of NaN values
nan_indices = X_train[X_train['Age'].isnull()].index

# Generate random numbers for the missing ages
random_ages = np.random.randint(lower_bound, upper_bound + 1, size=len(nan_indices))

# Fill NaN values using .loc to avoid SettingWithCopyWarning
X_train.loc[nan_indices, 'Age'] = random_ages

# Convert the Age column to integer data type
X_train['Age'] = X_train['Age'].astype(int)

print(f"\nMissing values in Age column after imputation: {X_train['Age'].isnull().sum()}")
print("\nX_train Age column after imputation and type conversion:")
display(X_train['Age'].head())
display(X_train['Age'].dtype)
```

### 3.10 Create a single function for preprocessing the test set (X_test) and apply it. (4 Marks)
#### **Note**: All the pre-processing steps that were applied on the train set before ML Modelling are also applied on the test set before passing through the predict function.

```
## Create a function
```

```
def preprocess_data(df):
    df_processed = df.copy()

    # 1. Cabin: Create 'Has_cabin' column
    # Ensure 'has_cabin' function is defined in the global scope
    df_processed['Has_cabin'] = df_processed['Cabin'].apply(has_cabin)

    # 2. Family Size: Create 'family_size' column
    df_processed['family_size'] = df_processed['SibSp'] + df_processed['Parch'] + 1

    # 3. Embarked: Fill missing values and map to numerical
    # Ensure 'most_frequent_embarked' and 'embarked_mapping' are defined globally
    df_processed['Embarked'].fillna(most_frequent_embarked, inplace=True)
    df_processed['Embarked'] = df_processed['Embarked'].map(embarked_mapping)

    # 4. Age: Impute missing values and convert to int
    # Ensure 'lower_bound' and 'upper_bound' are defined globally
    nan_age_indices = df_processed[df_processed['Age'].isnull()].index
    if not nan_age_indices.empty:
        random_ages = np.random.randint(lower_bound, upper_bound + 1, size=len(nan_age_indices))
        df_processed.loc[nan_age_indices, 'Age'] = random_ages
    df_processed['Age'] = df_processed['Age'].astype(int)

    # 5. Sex: Map to numerical and convert to int
    # Ensure 'sex_mapping' is defined globally
    df_processed['Sex'] = df_processed['Sex'].map(sex_mapping)
    df_processed['Sex'] = df_processed['Sex'].astype(int)

    # 6. Title: Extract, group rare, map to numerical, and convert to int
    # Ensure 'title_mapping' is defined globally
    df_processed['Title'] = df_processed['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
    df_processed['Title'] = df_processed['Title'].replace(['Lady', 'Countess','Capt', 'Col', \
         'Don', 'Dr', 'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
    df_processed['Title'] = df_processed['Title'].replace('Mlle', 'Miss')
    df_processed['Title'] = df_processed['Title'].replace('Ms', 'Miss')
    df_processed['Title'] = df_processed['Title'].replace('Mme', 'Mrs')
    df_processed['Title'] = df_processed['Title'].map(title_mapping)
    # Fill any titles not in the mapping (e.g., if a new rare title appears in test set)
    df_processed['Title'].fillna(title_mapping['Rare'], inplace=True)
    df_processed['Title'] = df_processed['Title'].astype(int)

    # 7. Fare Categorization: Categorize, map to numerical, and convert to int
    # Fill missing Fare values using the median from the training set to prevent data leakage
    if df_processed['Fare'].isnull().any():
        df_processed['Fare'].fillna(fare_median_train, inplace=True)

    # Use pd.qcut to categorize 'Fare'. This will use the current DataFrame's quantiles.
    # Ensure 'fare_mapping' is defined globally
    df_processed['Fare_Category'] = pd.qcut(df_processed['Fare'], 3, labels=['Low', 'Medium', 'High'])
    df_processed['Fare_Category'] = df_processed['Fare_Category'].map(fare_mapping)
    df_processed['Fare_Category'] = df_processed['Fare_Category'].astype(int)

    # 8. Drop Columns
    # Ensure 'columns_to_drop' is defined globally
    df_processed = df_processed.drop(columns=columns_to_drop)

    return df_processed
```

```
# Apply the preprocessing function to X_test
X_test_processed = preprocess_data(X_test)

print('X_test after preprocessing:')
display(X_test_processed.head())
display(X_test_processed.info())
```

## Exercise  4. Apply Multiple ML Algo. along with  Ensemble Technique (Voting classifier) and display the accuracy (7 Marks)
#### Expected Accuracy >= 80%  


```
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, VotingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Initialize individual classifiers
log_reg = LogisticRegression(random_state=42, solver='liblinear')
dec_tree = DecisionTreeClassifier(random_state=42)
rand_forest = RandomForestClassifier(random_state=42)
knn_classifier = KNeighborsClassifier()
svm_classifier = SVC(probability=True, random_state=42)
```

```
# Training and Evaluating Individual Classifiers
classifiers = {
    'Logistic Regression': log_reg,
    'Decision Tree': dec_tree,
    'Random Forest': rand_forest,
    'K-Nearest Neighbors': knn_classifier,
    'Support Vector Machine': svm_classifier
}

model_accuracies = {}

for name, classifier in classifiers.items():
    print(f"Training {name}...")
    classifier.fit(X_train, y_train)
    y_pred = classifier.predict(X_test_processed)
    accuracy = accuracy_score(y_test, y_pred)
    model_accuracies[name] = accuracy
    print(f"{name} Accuracy: {accuracy:.4f}\n")
```

```
# Creating and Evaluating the Voting Classifier (Ensemble Model)
# Create a Voting Classifier with a selection of models
voting_clf = VotingClassifier(
    estimators=[
        ('lr', log_reg),
        ('dt', dec_tree),
        ('rf', rand_forest),
        ('knn', knn_classifier),
        ('svm', svm_classifier)
    ],
    voting='soft' # 'soft' uses predicted probabilities, 'hard' uses predicted class labels
)

print("Training Voting Classifier...")
voting_clf.fit(X_train, y_train)
y_pred_ensemble = voting_clf.predict(X_test_processed)
ensemble_accuracy = accuracy_score(y_test, y_pred_ensemble)
model_accuracies['Voting Classifier'] = ensemble_accuracy
print(f"Voting Classifier Accuracy: {ensemble_accuracy:.4f}\n")
```

```
#Summary of Model Accuracies
print("\n--- Model Accuracies ---")
for name, accuracy in model_accuracies.items():
    print(f"{name}: {accuracy:.4f}")
```
