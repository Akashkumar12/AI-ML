```python
import sympy as sp
lam = sp.symbols('lambda')

expr = (
    (2.8-lam) *((5.2-lam)*(0.7-lam)-(0.3*0.3))
    -3.4 * (3.4*(0.7-lam) - (0.3)*(-0.4))
    +(-0.4) *(3.4*0.3-(5.2-lam)*(-0.4))
)
```

```python
expanded = sp.expand(expr)
expanded
```

```python
eigenvalues = sp.solve(expanded,lam)
eigenvalues
for val in eigenvalues:
    print(float(val.evalf().as_real_imag()[0]))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    "H":[126,128,128,130,130],
    "W":[78,80,82,82,84],
    "BMI":[49,48,50,48,49],
    "Gender":[1,1,0,0,1]
})
```

```python
print("Original Data")
print(df)
```

```python
df.columns
```

```python
X = df[['H', 'W', 'BMI']]
y = df['Gender']
```

```python
fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(1,1,1,projection="3d")
ax.set_xlabel("H")
ax.scatter(df['H'],df['W'],df['BMI'],c=df['Gender'],s=80)
plt.show()
```

```python
model = Pipeline([
          ("scaler",StandardScaler()),
           ("pca",PCA(n_components=2)),
          ("logistic",LogisticRegression())

          ])
```

```python
model.fit(X,y)
```

```python
X_scaled = model.named_steps['scaler'].transform(X)
X_scaled
```

```python
X_pca = model.named_steps['pca'].transform(X_scaled)
X_pca
```

```python
pca_df = pd.DataFrame(X_pca,columns=['PC1','PC2'])
pca_df
```

```python
plt.figure(figsize=(8,8))
plt.scatter(pca_df['PC1'],pca_df['PC2'],c=df['Gender'],s=80)
plt.show()
```

```python
pca = model.named_steps['pca']
print(pca.components_)
```

```python
pca.explained_variance_ratio_
```

```python
model.predict([[126 ,78 , 49]])
```
